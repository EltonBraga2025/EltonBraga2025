```python
import os
import sys

def calcular_e_salvar_media():
    """
    Pede notas, calcula a média e tenta salvar as notas e a média em um arquivo.
    """
    while True:
        try:
            # 1. Pedir ao usuário para inserir a primeira nota
            nota1_str = input("Por favor, insira a primeira nota (ou 'sair' para encerrar): ")
            if nota1_str.lower() == 'sair':
                return False # Sinaliza para sair do loop principal

            # 2. Armazenar a primeira nota em uma variável (e converter para float)
            nota1 = float(nota1_str)

            # 3. Pedir ao usuário para inserir a segunda nota
            nota2_str = input("Por favor, insira a segunda nota: ")

            # 4. Armazenar a segunda nota em outra variável (e converter para float)
            nota2 = float(nota2_str)

            break # Sai do loop de entrada se as notas forem válidas

        except ValueError:
            print("Entrada inválida. Por favor, insira apenas números para as notas.")
            # Continua no loop para pedir as notas novamente

    # 5. Calcular a média das duas notas
    media = (nota1 + nota2) / 2

    # 6. Exibir o resultado da média na tela
    print(f"A média das duas notas é: {media:.2f}") # .2f para formatar com 2 casas decimais

    # --- Parte de "Gravar" ---
    nome_arquivo = "notas_salvas.txt"
    try:
        with open(nome_arquivo, 'a') as arquivo: # 'a' para append (adicionar ao final do arquivo)
            arquivo.write(f"Notas: {nota1}, {nota2} - Média: {media:.2f}\n")
        print(f"Notas e média salvas em '{nome_arquivo}'.")
    except IOError as e:
        print(f"Erro ao tentar salvar as notas no arquivo: {e}")

    return True # Sinaliza que a operação foi bem-sucedida

# --- Parte de "Reiniciar" (Executar o programa repetidamente) ---
def main_menu():
    """
    Função principal que gerencia o loop de execução do programa.
    """
    while True:
        if not calcular_e_salvar_media():
            print("Programa encerrado.")
            break

        continuar = input("\nDeseja calcular outra média? (s/n): ").lower()
        if continuar != 's':
            print("Programa encerrado.")
            break
if __name__ == "__main__":
    main_menu()
```
